```python
import pandas as pd
```

## 1️⃣ Create a DataFrame
Create a DataFrame with columns: Name, Age, and Marks
Include 4 student records with at least one missing value.
(Hint: Use a dictionary and pd.DataFrame())

```python
data = {
    'Name':['Antony','Arun','Meera','John'],
    'Age' :[23,24,None,23],
    'Marks':[85,98,89,90]
}

df = pd.DataFrame(data)
print(df)
```

## 2️⃣ Add a new column
Add a new column "Grade" with letter grades (A, B, C, etc.)

```python
df['Grade'] = ['A', 'A+', 'B', 'A']
print(df)
```

## 3️⃣ Access specific columns
Display only the 'Name' and 'Marks' columns.

```python
print(df[['Name', 'Marks']])
```

## 4️⃣ Count missing values
Count the number of missing values in each column.
(Hint: df.isnull().sum())

```python
print(df.isnull().sum())
```

## 5️⃣ Handle missing values
Replace all missing values in 'Age' with the mean age.
(Hint: df['Age'].fillna(df['Age'].mean(), inplace=True))


```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

## 6️⃣ Modify a specific cell
Change the Name in the 3rd row (index 2) to 'Sonu'.
(Hint: df.loc[2, 'Name'] = 'Sonu')

```python
df.loc[2, 'Name'] = 'Sonu'
print(df)
```

## 7️⃣ Add new rows to the DataFrame
Add two new rows:
{'Name': 'Kumar', 'Age': 20, 'Marks': 90}
{'Name': 'Rahul', 'Age': 26, 'Marks': 85}
(Hint: Use pd.concat)

```python
new_rows = pd.DataFrame([
    {'Name': 'Kumar', 'Age': 20, 'Marks': 90, 'Grade': 'A'},
    {'Name': 'Rahul', 'Age': 26, 'Marks': 85, 'Grade': 'A'}
])

df = pd.concat([df, new_rows], ignore_index=True)
print(df)
```

## 8️⃣ Access data using loc
Access and print 'Name' and 'Marks' for the student at index 0.
(Hint: df.loc[0, ['Name', 'Marks']])

```python
print(df.loc[0, ['Name', 'Marks']])
```

## 9️⃣ Slicing rows and columns using loc
Display 'Name' and 'Marks' for rows 0 to 3.
(Hint: df.loc[0:3, ['Name', 'Marks']])

```python
print(df.loc[0:3, ['Name', 'Marks']])
```

## 🔟 Access rows and columns using iloc
Display the first and third rows, and first two columns.
(Hint: df.iloc[0:3:2, 0:2])


```python
print(df.iloc[0:3:2, 0:2])
```

## 1️⃣1️⃣ Aggregate functions
Calculate and display:
- Mean Age
- Minimum Age
- Maximum Age
- Count of Age values


```python
print("Mean Age:", df['Age'].mean())
print("Minimum Age:", df['Age'].min())
print("Maximum Age:", df['Age'].max())
print("Count of Age values:", df['Age'].count())
```

## 1️⃣2️⃣ Grouping and aggregation
Group the DataFrame by 'Name' and compute the mean Age for each Name.
(Hint: df.groupby('Name')['Age'].mean())

```python
print(df.groupby('Name')['Age'].mean())
```
