## 1️⃣ Series from a list with filtering


```python
import pandas as pd

s1 = pd.Series([10, 20, 30, 40, 50])

val_loc = s1.loc[3]
print(f"Element at index 3 (loc): {val_loc}")

val_iloc = s1.iloc[1]
print(f"Second element (iloc): {val_iloc}")

print("Elements > 25:")
print(s1[s1 > 25])
```

## 2️⃣ Series from a dictionary


```python
import pandas as pd

data_dict = {"A": 5, "B": 15, "C": 25, "D": 35}
s2 = pd.Series(data_dict)

val_b = s2.loc["B"]
print(f"\nValue for key 'B' (loc): {val_b}")

val_first = s2.iloc[0]
print(f"First element (iloc): {val_first}")

print("Values > 20:")
print(s2[s2 > 20])
```

## 3️⃣ DataFrame from a dictionary


```python
import pandas as pd

df = pd.DataFrame(
    {
        "Name": ["Alice", "Bob", "Charlie"],
        "Age": [25, 30, 35],
        "City": ["Delhi", "Mumbai", "Chennai"],
    }
)

print("\nEntire DataFrame:")
print(df)

print("\nRow at index 1 (loc):")
print(df.loc[1])

print("\nFirst row (iloc):")
print(df.iloc[0])
```

## 4️⃣ Conditional filtering in DataFrame


```python
import pandas as pd

age_filtered = df[df["Age"] > 28]
print("\nRows where Age > 28:")
print(age_filtered)

city_filtered = df[df["City"] == "Delhi"]
print("\nRows where City is 'Delhi':")
print(city_filtered)
```
