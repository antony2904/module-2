```python
import pandas as pd

df = pd.read_csv("TV_Final.csv")


df.columns = df.columns.str.strip()
```

## 1️⃣ Filtering based on condition
Q1: Display all rows where the Brand is "SAMSUNG" AND the Selling Price is less than 50000.

```python
lst = df[(df["Brand"] == "SAMSUNG") & (df["Selling Price"] < 50000)]
print(lst)
```

## 2️⃣ Conditional filtering with multiple conditions
Q2: Display only the Brand, Size, and Rating columns for HD LED where the Operating System is "Android" AND Rating is greater than or equal to 4.0. (Replaced laptops with HD LED)

```python
condition = (
    (df["Resolution"] == "HD LED")
    & (df["Operating System"].str.strip() == "Android")
    & (df["Rating"] >= 4.0)
)

q2 = df.loc[condition, ["Brand", "Size", "Rating"]]
print(q2)
```

## 3️⃣ Drop rows with missing values
Q3: Drop all rows where the Rating column has NaN values.

```python
df_cleaned = df.dropna(subset=["Rating"])
print(df_cleaned)
```

## 4️⃣ Fill missing values
Q4: Replace all NaN values in the Operating System column with the string "Unknown".

```python
df["Operating System"] = df["Operating System"].fillna("Unknown")
print(df["Operating System"].value_counts(dropna=False))
```

## 5️⃣ Rename columns
Q5: Rename the column "Selling Price" to "Current Price" and "Original Price" to "MRP".

```python
df = df.rename(columns={"Selling Price": "Current Price", "Original Price": "MRP"})
print(df.head())
```

## 6️⃣ Count occurrences using [value_counts()]
Q6: Find how many HD LED are available for each Brand using value_counts(). (Replaced laptops with HD LED)

```python
hd_led_counts = df[df["Resolution"] == "HD LED"]["Brand"].value_counts()
print(hd_led_counts)
```

## 7️⃣ Grouping and aggregation
Q7: Group the DataFrame by Brand and calculate the sum of Selling price for each brand. Then, sort the result in descending order of average selling price.

```python
price_col = "Current Price" if "Current Price" in df.columns else "Selling Price"

lst = (
    df.groupby("Brand")
    .agg(
        Total_Selling_Price=(price_col, "sum"),
        Avg_Selling_Price=(price_col, "mean"),
    )
    .sort_values(by="Avg_Selling_Price", ascending=False)
)

print(lst)
```

## 8️⃣ Sorting values
Q9: Sort the DataFrame first by Selling Price in ascending order and then by Rating in descending order.

```python
price_col = "Current Price" if "Current Price" in df.columns else "Selling Price"

lst = df.sort_values(by=[price_col, "Rating"], ascending=[True, False])
print(lst)
```

## 9️⃣ Groupby with sort
Q10: Find the top 3 brands with the highest selling price. Display their brand names.

```python
price_col = "Current Price" if "Current Price" in df.columns else "Selling Price"

top_3_brands = (
    df.groupby("Brand")[price_col].max().nlargest(3).index.tolist()
)
print("Top 3 brands with highest selling price:", top_3_brands)
```
