## 1️⃣ LINE GRAPH


Q1: Create a line graph to show the population growth of India from 2000 to 2020.
- years = [2000, 2005, 2010, 2015, 2020]
- population = [1.0, 1.1, 1.2, 1.3, 1.38]
- Add labels for x and y axes
- Add title "Population Growth of India (2000–2020)"
- Display grid lines

```python
import matplotlib.pyplot as plt

years = [2000, 2005, 2010, 2015, 2020]
population = [1.0, 1.1, 1.2, 1.3, 1.38]

plt.plot(years, population, marker='o')

plt.xlabel("Year")
plt.ylabel("Population (in billions)")
plt.title("Population Growth of India (2000–2020)")
plt.grid(True)

plt.show()
```

Q4: Plot a grouped bar chart comparing male vs female students in each department.
- departments = ['CSE', 'ECE', 'EEE', 'ME']
- male = [40, 35, 30, 25]
- female = [30, 25, 20, 15]
- Display bars side by side
- Add a legend

```python
import matplotlib.pyplot as plt
import numpy as np

departments = ['CSE', 'ECE', 'EEE', 'ME']
male = [40, 35, 30, 25]
female = [30, 25, 20, 15]

x = np.arange(len(departments))
width = 0.35

plt.bar(x - width/2, male, width,
        label='Male', color='red')

plt.bar(x + width/2, female, width,
        label='Female', color='blue')

plt.xlabel("Departments")
plt.ylabel("Number of Students")
plt.title("Male vs Female Students")
plt.xticks(x, departments)
plt.legend()

plt.show()
```

Q10: Generate a histogram for 100 random numbers normally distributed around mean = 50 and std = 10 using NumPy.
- Plot histogram with 10 bins
- Add grid and color customization

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

data = np.random.normal(loc=50, scale=10, size=100)

plt.hist(data, bins=10, color='red', edgecolor='black')

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normally Distributed Random Numbers")
plt.grid(axis='y', linestyle='--')

plt.show()
```

## 1️⃣ BAR GRAPH: Total Quantity Sold by Category

Task: Create a bar graph to show the total quantity sold for each product category.
Hint:
1. Group by 'Category' and sum 'Quantity'
2. Use plt.bar() to plot
3. Label X-axis, Y-axis, and add a title
Example:
- category_quantity = df.groupby('Category')['Quantity'].sum()
- plt.bar(category_quantity.index, category_quantity.values)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('orders.csv')
```

```python
category_quantity = df.groupby('Category')['Quantity'].sum()

plt.bar(category_quantity.index, category_quantity.values,
        color='skyblue', edgecolor='black')

plt.xlabel('Category')
plt.ylabel('Total Quantity Sold')
plt.title('Total Quantity Sold by Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 2️⃣ BAR GRAPH: Total Sales by Country


Task: Create a bar chart showing total sales for each country.
Hint:
1. Add a 'Sales' column: df['Sales'] = df['Quantity'] * df['Price']
2. Group by 'Country' and sum 'Sales'
3. Sort values in descending order
4. Use plt.bar() to plot
5. Display the sales value above each bar

```python
df['Sales'] = df['Quantity'] * df['Price']

country_sales = df.groupby('Country')['Sales'].sum()
country_sales = country_sales.sort_values(ascending=False)

plt.bar(country_sales.index, country_sales.values,
        color='orange', edgecolor='black')

plt.xlabel('Country')
plt.ylabel('Total Sales')
plt.title('Total Sales by Country')
plt.xticks(rotation=45)

for i, value in enumerate(country_sales.values):
    plt.text(i, value, f'{value:.0f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

## 3️⃣ LINE GRAPH: Daily Sales Trend


Task: Plot a line graph showing total sales per day.
Hint:
1. Convert 'OrderDate' to datetime: df['OrderDate'] = pd.to_datetime(df['OrderDate'])
2. Group by 'OrderDate' and sum 'Sales'
3. Use plt.plot() with markers and line style
4. Label X-axis, Y-axis, and add a title

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
df['Sales'] = df['Quantity'] * df['Price']

daily_sales = df.groupby('OrderDate')['Sales'].sum()
daily_sales = daily_sales.sort_index()

plt.plot(daily_sales.index, daily_sales.values,
         color='green', marker='o', linestyle='-')

plt.xlabel('Order Date')
plt.ylabel('Total Sales')
plt.title('Daily Sales Trend')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

# 5️⃣ Create a subplot with two plots side by side:
- Left plot: Bar chart of 'Product Category' vs 'Quantity Sold'.
- Right plot: Pie chart showing 'Category' wise sales percentage.

Dataset 5: Category and Quantity


```python
df_sales = pd.DataFrame({
    'Product Category': ['Electronics', 'Clothing', 'Groceries',
                         'Toys', 'Books'],
    'Quantity Sold': [120, 200, 150, 90, 130]
})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Left plot: Bar chart
axes[0].bar(
    df_sales['Product Category'],
    df_sales['Quantity Sold']
)
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Quantity Sold")
axes[0].set_title("Product Category vs Quantity Sold")
axes[0].tick_params(axis='x', rotation=30)

# Right plot: Pie chart
axes[1].pie(
    df_sales['Quantity Sold'],
    labels=df_sales['Product Category'],
    autopct='%1.1f%%',
    startangle=90
)
axes[1].set_title("Category-wise Sales Percentage")

plt.tight_layout()
plt.show()
```

# 6️⃣ Create a 2x1 subplot layout:

- Bottom plot: Scatter plot showing 'Advertising Spend' vs 'Revenue'.

```python
df_company = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021, 2022, 2023],
    'Revenue': [50000, 60000, 70000, 85000, 95000, 110000],
    'Ad Spend': [5000, 7000, 9000, 10000, 11000, 13000]
})
```

```python
fig, axes = plt.subplots(2, 1, figsize=(8, 10))

axes[0].plot(df_company['Year'], df_company['Revenue'],
             marker='o')
axes[0].set_xlabel("Year")
axes[0].set_ylabel("Revenue")
axes[0].set_title("Yearly Revenue")
axes[0].grid(True)

axes[1].scatter(df_company['Ad Spend'], df_company['Revenue'],
                marker='o')
axes[1].set_xlabel("Advertising Spend")
axes[1].set_ylabel("Revenue")
axes[1].set_title("Advertising Spend vs Revenue")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
