## Matplotlib

# Data visulization using python 2D plotting library

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np




print(matplotlib.__version__)
```

```python
x= np.array([2022,2023,2024,2025])
y = np.array([0,0,14,25])


plt.plot(x,y, marker ='o',
        markersize = 8,
        markerfacecolor = 'Red',
        markeredgecolor = 'Red',
        linestyle = 'solid',
        linewidth = 2,
        color ='Green' )

plt.show()
```

```python
x = np.array([2022,2023,2024,2025])

y1 = np.array([2,3,5,7])
y2 = np.array([5,2,7,4,])
y3 = np.array([34,56,23,98])
y4 = np.array([3,56,2,77])

line_style = dict(marker = 'o',
                  markersize = 10,
                  markerfacecolor = 'Red',
                  markeredgecolor = 'Red', 
                  linestyle = 'solid', 
                  linewidth = 2, 
                  color ='Blue')

plt.plot(x ,y1, **line_style)
plt.plot(x,y2, **line_style )
plt.plot(x,y3, **line_style )
plt.plot(x,y4, **line_style)

plt.show
```

'''
 grid() = grid() function in Matplotlib is used to enhance the readability and interpretability 
 of plots by adding a grid of horizontal and/or vertical lines. These lines serve as visual aids, 
 making it easier to analyze and understand the data presented in a graph.
'''

```python
x = np.array([1,2,3,4,5])
y = np.array([5,10,15,20,25])

plt.grid(linewidth = 1,
         color = 'Red',
         linestyle = ':')

plt.plot(x,y, marker = 'v')
plt.show()
```

```python
fruits = np.array(['Apple','Bananas','cherries','Dates'])
sales = np.array([500,300,650,750])

line_style = dict(fontsize = 18,
                 family = 'Arial',
                 fontweight = 'bold',
                 color = 'Blue')

plt.bar(fruits ,sales,color = 'skyblue')
plt.title('Fruits Sales',fontsize = 18,color = 'skyblue')
plt.xlabel('fruits',**line_style)
plt.ylabel('sales',**line_style)

plt.show
```

## Pie Chart

```python
x = np.array([25,50,5,8,67])

mylabels = ["Apples","Orange","Bananas","Cherry","Mangoes"]

plt.pie(x,labels = mylabels,
        autopct = "%1.2f%%",)

plt.title('Furits Sales',fontsize = 18,color = 'red')
plt.show()
```

## Scatter Port

```python
x1 = np.array([1,2,4,6,8,7,5,9,0,6])
y1 = np.array([10,20,30,45,50,75,80,90,45,67])

plt.title('Test Scores')
plt.xlabel('Hours studied')
plt.ylabel('Marks')

plt.scatter(x1,y1)
plt.show()
```

## Histogram

```python
x = np.random.normal(loc = 80 , scale = 15 , size =150)

plt.hist(x)
plt.show()
```

## Sub Plots

```python
x = np.array([1,2,3,4,5])

figure ,axes = plt.subplots(2,2)

axes[0,0].plot(x,x*2,color='Red')
axes[0,0].set_title('multiple_X')

axes[0,1].plot(x,x**2,color='Blue')
axes[0,1].set_title('X**2')

axes[1,0].plot(x,x**3,color='Green')
axes[1,0].set_title('X**3')

axes[1,1].plot(x,x**4,color='Yellow')
axes[1,1].set_title('X**4')

plt.tight_layout()
plt.show()
```
